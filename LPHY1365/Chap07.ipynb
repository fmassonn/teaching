Avec la température et les précipitations, le vent est assurément un des paramètres météorologiques qui nous importent le plus au quotidien. Invisible mais pourtant omniprésent, il peut en même temps transporter les sables du Sahara jusqu’à nous, disperser les pollens qui chatouillent nos nez lors du printemps, et même parfois arracher les tuiles que l’on croyait solidement fixées sur notre toi. Mordant lorsqu’il descend du nord en hiver, brûlant et sec lorsqu’il remonte du sud en été, il module l’évaporation des sols et l'évapotranspiration des plantes, influençant discrètement mais profondément les équilibres hydrologiques.

Les sociétés humaines ont très tôt appris à composer avec lui — et même à l’exploiter — en dressant des moulins face à sa course afin de profiter de son action, ou en l'utilisant pour se déplacer sur les mers. Aujourd'hui, le vent est une donnée centrale pour la planification énergétique et la transition vers un monde bas carbone.

Mais au fond, qu’est-ce que le vent ? D’où vient-il, pourquoi souffle-t-il, et quelles forces invisibles mettent l’air en mouvement à l’échelle d’une personne, d'une ville, ou d’un continent ?

Pour débuter ce chapitre, posons-nous quelques questions:

- Pourquoi l'air se met-il en mouvement? 
- Quelle(s) est (sont) la (les) cause(s) du vent?
- Du vent serait-il observé sur une Terre sans rotation?
- Les vents sont-ils plus forts en altitude ou au sol? Pourquoi?
- Les vents sont-ils plus forts au voisinage des basses ou des hautes pressions?
- Toutes choses étant égales par ailleurs, le vent est-il plus fort dans une atmosphère plus chaude ou plus froide?
- Deux atmosphères ayant le même champ de pression en surface peuvent-elles avoir des vents différents?
- Quel est l'impact de la présence d'obstacles au sol pour le vent en surface?
- Les dynamiques sont-elles comparables à l'équateur, aux tropiques, aux moyennes latitudes, dans les régions polaires?




Dans ce chapitre, nous étudierons le vent à **l'échelle synoptique**. Signifiant "vue (*opsis*) d'ensemble (*sun*)", l'échelle synoptique est la dimension caractéristique des phénomènes atmosphériques qui permettent la description de l'ensemble de la dynamique atmosphérique planétaire, en ce compris les ondes planétaires, les dépressions et les anticyclones. Pour les mouvements horizontaux, la longueur caractéristique de l'échelle synoptique est donc de l'ordre du millier de kilomètres, et son temps caractéristique est de l'ordre de la journée. Pour les mouvements verticaux, nous travaillerons toujours sous l'hypothèse de l'équilibre hydrostatique et nous pourrons donc pas étudier, par exemple, les mouvements ascendants intenses qu'on trouve dans les tornades ou les ouragans, par exemple.

Le vent à l'échelle synoptique ne se mesure donc pas directement, mais est plutôt estimé en moyennant dans le temps des mesures collectées à haute fréquence (minutes) afin de lisser les fluctuations dues aux dynamiques turbulentes, de micro-échelle, aérologiques et de méso-échelle.

## 7.1 - Equation générale du mouvement

On rappelle ici qu'on travaille dans une repère orthonormé $\left(\vec{i}, \vec{j}, \vec{k}\right)$ désignant les directions "est, nord, et haut" respectivement. Ce repère est entraîné par la rotation solide de la Terre à raison d'une vitesse angulaire $\Omega = 2\pi / 86 400 \textrm{s} \approx 7.27 \times 10^{-5} \textrm{ s}^{-1}$. 

Au cours du chapitre précédent, nous avons démontré que la seconde loi de Newton pour une parcelle d'air de masse $m$ pouvait s'exprimer, dans ce repère en rotation, par la relation

$$
\sum \vec{F}_\textrm{réelles} -  m f \vec{k} \times\vec{u} = m\vec{a}
$$

où $f = 2\Omega\sin\phi$ est le paramètre de Coriolis et 

$$
     \vec{a} := \frac{\textrm{d}\vec{u}}{\textrm{d}t} 
$$

désigne l'accélération, mesurée dans ce repère tournant, de la parcelle étudiée.

Avant de pouvoir résoudre cette équation pour $\vec{a}$ et, *in fine*, pour le champ de vitesse $\vec{u}$, nous devons spécifier l'ensemble des forces réelles impliquées dans cette équation.

### 7.1.1 - Les forces en présence

Imaginons-nous un instant être  à la place de cette parcelle d'air, et demandons-nous quelle(s) forces réelles nous ressentirions alors dans cette situation. Plus précisément, déterminons le module et la direction de ces forces.

Trois forces réelles entrent en jeu:


* Le poids

Le poids s'exprime donc $\vec{\Pi} := -mg\vec{k}$ où $g$=9.81 m/s $^2$ est l'accélération de la gravité. En toute généralité, le poids dépend de l'altitude par rapport au sol $z$ et de la latitude $\phi$ en raison de la non-sphéricité de la Terre et de l'influence de la force centrifuge; nous négligerons ces variations ici.

* La résultante des forces de frottement du sol

Nous désignons cette force par $\vec{\mathcal{F}}:= -\mathcal{F}\frac{\vec{u}}{\| \vec{u}|}$: cette force s'oppose bien par nature au mouvement. La force de frottement (moléculaire) de l'air en altitude est en général considérée comme négligeable et est donc omise.

* La résultante des forces de pression

La pression est, on le rappelle, la force exercée sur une surface de 1 m $^2$ par les molécules d'un liquide ou d'un gaz qui sont en mouvement du fait de leur agitation thermique. Cette pression est d'autant plus grande que la quantité de mouvement communiquée à cette surface est élevée; ceci est le cas lorsque l'énergie cinétique moyenne des molécules (i.e., la température) est élevée, et que le nombre de molécules pouvant heurter la surface (proportionnel à la densité) est élevée: c'est la loi des gaz parfaits.

Dans une atmosphère statique, la pression est isotrope: la force exercée par les molécules en mouvement est identique dans toutes les directions considérées.

On démontre au cours que la résultante des forces de pression s'appliquant sur une parcelle de masse $m$ est donnée par

$$ \vec{F}_p = -\frac{m}{\rho}\nabla p \qquad \textrm{("Force de gradient de pression")}$$

avec 
$$
\nabla p := \left(\frac{\partial p}{\partial x}, \frac{\partial p}{\partial y}, \frac{\partial p}{\partial z}\right) \qquad \textrm{("Gradient de pression")}
$$

Nous travaillerons à partir de maintenant avec une parcelle d'air de masse unitaire $m=1$ kg (ou, de façon équivalente, nous exprimons les équations du mouvement par unité de masse). Il vient donc :

\frac{\textrm{d}\vec{u}}{\textrm{d}t}  = - f\vec{k}\times\vec{u} -\frac{1}{\rho}\nabla p -g\vec{k} + \vec{\mathcal{F}} \qquad \textrm{(Equation générale du mouvement)} 
$$




### 7.1.2 - L'équation du mouvement selon les axes

L'équation (vectorielle) générale du mouvement ne peut être résolue telle quelle; il est nécessaire de l'exprimer sur chacun des trois axes principaux $(\vec{i}, \vec{j}, \vec{k})$ correspondant aux coordonnées $(x, y, z)$ du mouvement et aux composantes $(u, v, w)$ du champ de vitesse. En supposant pour l'instant que les forces de frottement du sol ont un effet négligeable (ce qui revient à nous placer dans l'atmosphère libre, au-delà de 1500-2000 m d'altitude), et par développement du produit vectoriel, on obtient alors un système de trois équations:

$$
\left\{
\begin{aligned}
\frac{\textrm{d}u}{\textrm{d}t} &= fv &-\frac{1}{\rho}\frac{\partial p}{\partial x} \\
\frac{\textrm{d}v}{\textrm{d}t} &= -fu &-\frac{1}{\rho}\frac{\partial p}{\partial y} \\
\frac{\textrm{d}w}{\textrm{d}t} &= -g  & -\frac{1}{\rho}\frac{\partial p}{\partial z}
\end{aligned}
\right.
$$ 

Opérons maintenant une distinction entre la dynamique verticale de la dynamique horizontale.

## 7.2 - Equation du mouvement selon la verticale

La troisième composante de l'équation du mouvement selon les axes s'exprime

$$ 
\frac{\textrm{d}w}{\textrm{d}t} =  -g   -\frac{1}{\rho}\frac{\partial p}{\partial z}
$$

L'examen des ordres de grandeur des trois termes impliqués révèle qu'à l'échelle synoptique, seuls les deux termes du membre de droite sont prépondérants:
* $g$ est de l'ordre de 10 m/s²
* A 700 hPa, dans l'atmosphère libre, les variations verticales de pression observées sont de l'ordre de 1 hPa / 10 m, et à une température de -5°C, l'air (sec) y a une densité de 0.9 kg/m³. La force de gradient de pression est donc de l'ordre de 10 m/s² également
* Le terme d'accélération verticale est, à l'échelle synoptique, plusieurs ordres de grandeur inférieures à ces deux termes.  

L'équation du mouvement selon la verticale peut donc s'écrire:

$$ 
0 =  -g   -\frac{1}{\rho}\frac{\partial p}{\partial z}
$$

ce qui n'est autre que la relation de l'équilibre hydrostatique déjà rencontrée au Chapitre 5.

### 7.2.1 - La coordonnée pression

En météorologie (comme de façon plus générale, en sciences de l'atmosphère ou en aéronautique), la coordonnée verticale $z$ est rarement utilisée en pratique. Plusieurs raisons expliquent ce choix:

- la dynamique atmosphérique est étroitement liée à la configuration des surfaces isobares, comme nous allons le voir ci-après. Un système de coordonnées reflétant cette réalité est donc plus approprié;
- la pression est directement mesurée au moyen d'instruments tels que des baromètres embarqués dans des ballons-sondes, alors que l'altitude requiert des technologies plus avancées (radar, ...);
- les équations du mouvement s'expriment de façon plus compacte et simple en coordonnées pression qu'en coordonnées $z$.

L'utilisation de la coordonnée pression est rendue possible en lieu et place de la coordonnée $z$ en raison du fait qu'une correspondance bijective existe toujours, à l'échelle synoptique, entre $z$ et $p$. Cette relation change certes en fonction de la densité de l'air, mais elle est unique en tout point du globe. On peut donc légitimement et sans ambiguïté, décrire la hauteur d'une parcelle au moyen de sa pression, en traitant alors $z$ comme une variable dépendante.


#### 7.2.1.1 - La force horizontale de gradient de pression en coordonnées pression

L'équation générale du mouvement implique, selon l'horizontale, un terme $-\frac{1}{\rho}\nabla_h p$ avec 
$$\nabla_h p := \left(\frac{\partial p}{\partial x}, \frac{\partial p}{\partial y}\right)$$

Afin de travailler en coordonnées pression, nous devons donc exprimer ces deux termes en prenant $p$, et non $z$, comme la coordonnée verticale indépendante.

On démontre au cours que le gradient horizontal du champ de pression devient, en coordonnées pression,

$$ 
\nabla_h p = \rho g \nabla_p z = \rho G \nabla_p Z  = \rho  \nabla_p \Phi \qquad \textrm{("Gradient isobare de géopotentiel")}
$$

où on a utilisé la définition de variation de géopotentiel $\textrm{d}\Phi = g \textrm{d}z = G \textrm{d}Z$, permettant d'absorber les variations de $g$ avec l'altitude dans une variable $Z$ mesurée en mètres géopotentiels (mgp) et $G=9.8$ m/s² une constante (vue au Chapitre 5).

On s'intéressera donc rarement, à partir de maintenant à l'altitude des surfaces isobares; par contre, on s'intéressera de près aux surfaces d'égal géopotentiel pour un niveau de pression donné. Ces surfaces sont appelées "isohypses" (du grec "iso" signifiant *égal* et "hypsos" signifiant "hauteur").

Les composantes horizontales de la force de gradient de pression peuvent, grâce aux relations établies ici, s'exprimer de façon beaucoup plus compacte qu'en coordonnées $z$:

$$
\vec{F}_{p,h} = -\frac{1}{\rho} \nabla_h p = -\nabla_p \Phi
$$

Il suffit donc de connaître la distribution spatiale du champ de géopotentiel $\Phi$ pour en déduire les composantes horizontales de la force de gradient de pression.

#### 7.2.1.2 - La vitesse verticale en coordonnées pression

L'utilisation de coordonées pression nous permet également redéfinir la notion de vitesse verticale. Par définition, la vitesse verticale $w$ "métrique" (mesurée en mètres parcourus verticalement par seconde) n'est autre que la variation, avec le mouvement, de la propriété "altitude" d'une parcelle que nous mesurons à l'aide de la variable $z$, prise en général comme une coordonnée indépendante. Nous avons vu au Chapitre 6 que la dérivée temporelle d'une propriété quelconque $X$ suivant le mouvement est donnée par la dérivée matérielle $\textrm{d}/\textrm{d}t$:


$$ 
\frac{\textrm{d}X}{\textrm{d}t} = \frac{\partial X}{\partial t} + \frac{\textrm{d}x}{\textrm{d}t} \frac{\partial X}{\partial x} + \frac{\textrm{d}y}{\textrm{d}t} \frac{\partial T}{\partial y} + \frac{\textrm{d}z}{\textrm{d}t} \frac{\partial T}{\partial z} 
$$

La vitesse verticale géométrique est donc, par définition,

$$
w:=\frac{\textrm{d}z}{\textrm{dt}} \qquad \textrm{Définition de vitesse verticale en coordonnées } z 
$$

ce qui donne bien, en appliquant la règle dérivée matérielle et en considérant que $x, y, z$ sont des coordonnées indépendantes,

$$ 
w = \frac{\partial z}{\partial t} + u \frac{\partial z}{\partial x}  + v \frac{\partial z}{\partial y} + w\frac{\partial z}{\partial z} = w  
$$


Lorsque nous faisons le choix de travailler en coordonnées pression plutôt qu'en coordonnées $z$, nous devons aussi redéfinir la notion de vitesse verticale puisque celle-ci ne peut plus s'exprimer en mètres parcourus verticalement par seconde (étant donné que $z$ n'est plus une coordonnée indépendante). En reprenant la définition de vitesse verticale en coordonnées $z$, nous pouvons alors définir la **vitesse verticale en coordonnées pression**:

$$ 
\omega := \frac{\textrm{d}p}{\textrm{dt}}  = \frac{\partial p}{\partial t} + u \frac{\partial p}{\partial x}  + v \frac{\partial p}{\partial y} + w\frac{\partial p}{\partial z} 
$$

Or, pour les mouvements synoptiques, $\partial p / \partial t \sim $ 10 hPa/jour, $u\ \partial p/\partial x, v\ \partial p/\partial y\sim$ (10 m/s).(1 Pa/km)~ 10 hPa/jour et $g\rho w\sim$ (10 m/s²).(1 kg/m³). (0.01 m/s)$\sim$ 0.001 Pa/s $\sim$ 100 hPa/jour. La vitesse verticale en coordonnées pression peut donc être approximée par 

$$
\omega = -\rho g w
$$

et se mesure en Pa/s; elle est négative pour les mouvements ascendants et positives pour les mouvements descendants. 

### 7.2.2 - Les équations d'évolution en coordonnées pression

Au Chapitre 6, dans la section 6.4, nous avions établi l'expression de la variation instantanée d'une propriété telle qu'elle serait perçue au sein d'une parcelle d'air évoluant dans un champ de vitesse connu. Cette expression peut, en toute rigueur, s'établir en coordonnées pression. Si $X=X(t,x,y,p)$ est un champ quelconque, alors par dérivée de fonctions composées,

$$ \frac{\textrm{d}X}{\textrm{d}t} = \left(\frac{\partial X}{\partial t}\right)_p + u \left(\frac{\partial X}{\partial x}\right)_p + v \left(\frac{\partial X}{\partial y}\right)_p + \omega \frac{\partial X}{\partial p} \qquad \textrm{(Dérivée matérielle en coordonnées pression)}$$

où on a utilisé le fait que par définition, $\omega = \textrm{d}{p}/\textrm{d}t$. Les indices $\left(\ \right)_p$ rappellent que nous travaillons à pression constante. 

Si on se place en un point fixe $x,y,p$, alors la variation observée de $X$ sera donnée par

$$
\underbrace{\left(\frac{\partial X}{\partial t}\right)_p}_{\textrm{tendance}}= 
 \underbrace{-\left[ u \left(\frac{\partial X}{\partial x}\right)_p
+ v \left(\frac{\partial X}{\partial y}\right)_p\right]}_{\text{advection isobare horizontale de } X} \
\underbrace{- \omega \frac{\partial X}{\partial p}}_{\text{advection verticale de $X$}} \ + \frac{\textrm{d}X}{\textrm{d}t}
$$

La variation de la propriété $X$ en un endroit fixé (sa tendance) est donc définie, d'une part par le transport à contre-gradient isobare de la propriété étudiée, et d'autre part par la variation intrinsèque de la propriété en raison des tranformations au sein de la parcelle.

## 7.3 - Le mouvement horizontal

Au vu des développements effectués dans la section précédente, l'équation générale du mouvement peut s'exprimer, en coordonnées pression,

$$
\left\{
\begin{aligned}
\frac{\textrm{d}\vec{u}_h}{\textrm{d}t} &= - f\vec{k}\times\vec{u}_h - \nabla_p \Phi \\
\frac{\textrm{d}w}{\textrm{d}t} & = -g   -\frac{1}{\rho}\frac{\partial p}{\partial z} 
\end{aligned}
\right.
$$ 


avec $\vec{u}_h := (u, v)$ le champ de vitesse horizontal. On voit ici tout l'avantage de travailler en coordonnées pression: le terme de densité a tout simplement disparu de la composante horizontale de l'équation du mouvement!


On réalise, en classe, une analyse d'ordre de grandeur des termes impliqués dans la composante horizontale de l'équation générale du mouvement. Il résulte de cette analyse que pour les échelles synoptiques et dans l'atmosphère libre, le terme d'accélération peut être ignoré. En posant $\textrm{d}\vec{u}_h/\textrm{d}t=0$, il vient alors que la dynamique horizontale résulte d'un équilibre entre deux forces: c'est **l'équilibre géostrophique**.


### 7.3.1 - Le vent géostrophique

L'équilibre géostrophique se définit donc par la relation 

$$
0 = - f\vec{k}\times\vec{u}_h - \nabla_p \Phi \qquad \textrm{(Equilibre géostrophique)}
$$

qui est un cas particulier de l'équation générale du mouvement dans l'horizontale, dans laquelle les termes d'accélération ont été négligés.


Il convient d'observer que cette relation est _vectorielle_ (deux équations s'y cachent) mais algébrique: l'inconnue $\vec{u}_h$ n'y apparaît pas sous forme de dérivée. 


La solution pour le champ de vitesse $\vec{u}_h$ qui satisfait cette relation est appelé "**vent géostrophique**". Découvrons maintenant, en trois temps, les propriétés de ce vent.

#### 7.3.1.1 - Approche graphique
On suppose connaître ce vent géostrophique en un point donné et on construit au fur et à mesure les deux forces (de Coriolis et de gradient isobare de géopotentiel) qui définissent l'équilibre dynamique à satisfaire:

**Il vient que dans l'hémisphère nord, le vent géostrophique souffle parallèlement aux isohypses en laissant les zones de  bas géopotentiels sur sa gauche.** Dans l'hémisphère sud, il laisse les bas géopotentiels sur sa droite.

#### 7.3.1.2 - Approche algébrique

En projetant sur les axes horizontaux la relation de l'équilibre géostrophique, on trouve que le vent géostrophique $\vec{u}_g^h = (u_g^h, v_g^h)$ doit satisfaire les relations:

$$
\left\{
\begin{aligned}
fv_h^g & = & \left(\frac{\partial \Phi}{\partial x}\right)_p  \\
-fu_h^g & = & \left(\frac{\partial \Phi}{\partial y}\right)_p  \\
\end{aligned}
\right.
$$ 

ou encore

$$
\left\{
\begin{aligned}
u_h^g & = & - \frac{1}{f}\left(\frac{\partial \Phi}{\partial y}\right)_p  \\
v_h^g & = & \frac{1}{f}\left(\frac{\partial \Phi}{\partial x}\right)_p  \\
\end{aligned}
\right.
$$ 

#### 7.3.1.2 - Approche vectorielle

L'expression obtenue pour le vent géostrophique peut encore se réécrire sous forme vectorielle à l'aide de la définition de produit vectoriel:

$$
\vec{u}_h^g = \frac{1}{f}\vec{k}\times\nabla_p\Phi
$$


On remarque que le module du vent géostrophique est donné par 
$$
\| \vec{u}_h^g \| = \left\| \frac{1}{f} . \frac{\Delta\Phi}{\Delta l} \right\| = \left\| \frac{G}{f} . \frac{\Delta Z}{\Delta l} \right\|
$$

On voit donc que la vitesse du vent géostrophique est proportionnelle au gradient de géopotentiel $\Phi$, c'est-à-dire au resserrement des lignes isohypses.

### 7.3.3 - Le vent thermique

La section précédente, consacrée au rôle du frottement,  a illustré la manière dont le vent change en intensité et en direction en raison de la présence de forces de frottement au sol. Il se trouve que, même en l'absence de frottement, la distribution verticale du vent horizontal n'est en général pas homogène: nous allons voir que cette distribution dépend fondamentalement du champ de température. Le concept de **vent thermique**, que nous introduisons, fait apparaître un lien structurant entre la configuration dynamique et la structure thermodynamique de l'atmosphère.

On se rappelle (Chapitre 5) de la **formule de Laplace**:

$$
\Phi_2 - \Phi_1 := G(Z_2-Z_1) = R_a T_{vm} \ln\left( \frac{p_1}{p_2}\right)
$$

avec $(1,2)$ deux niveaux définissant une couche de pression comprise entre les isobares $p_1>p_2$, $\Phi$ le géopotentiel, $Z$ la hauteur de géopotentiel, $T_{vm}$ le température moyenne de la couche considérée, et $R_a$ la constante des gaz parfaits pour l'air sec. Cette relation nous enseigne qu'une couche de pression est d'autant plus dilatée que la température moyenne en son sein est élevée.

Considérons maintenant les vents géostrophiques à ces deux niveaux de pression $p_1$ et $p_2$, et supposons en toute généralité que les deux surfaces isobares correspondantes ne sont pas parallèles entre elles. Soient $\vec{u}_{h,1}^g$ et $\vec{u}_{h,2}^g$ les vents géostrophiques à ces deux niveaux de pression. Comme nous l'avons vu dans la section 7.3.1.2, le vent géostrophique est défini par le champ de géopotentiel par la relation

$$
\vec{u}_h^g = \frac{1}{f}\vec{k}\times\nabla_p\Phi
$$ 

Il vient donc que la différence de vent géostrophique entre les niveaux $2$ et $1$ est donnée par 

$$
\vec{u}_T := \vec{u}_{h,2}^g - \vec{u}_{h,1}^g = \frac{1}{f}\vec{k}\times\nabla_p\left(\Phi_2 - \Phi_1\right)
$$

ce qui, en vertu de la relation de Laplace, donne

$$
\vec{u}_T := \vec{u}_{h,2}^g - \vec{u}_{h,1}^g = \frac{R_a}{f}\ln\left(\frac{p_1}{p_2}\right)\vec{k}\times\nabla_p\left( T_{vm} \right)
$$

Le vecteur "différence de vent géostrophique entre les niveaux $1$ et $2$ est appelé **vent thermique** et est symbolisé par l'expression $\vec{u}_T$. On voit via l'expression obtenue que le vent thermique "souffle" parallèlement aux isothermes en laissant les basses température sur sa gauche.

#### 7.3.3.1 - Renforcement avec l'altitude, des vents d'Ouest dans la troposphère, et des vents d'Est dans la stratosphère

Comme on l'a vu au Chapitre 1 du cours, le champ thermique moyen $T_{vm}$ est tel qu'on va du "froid" vers le chaud lorsqu'on se déplace de l'équateur vers les pôles. Il en résulte, en vertu de la relation de Laplace, que les couches comprises entre deux niveaux de pression prédéfinis ont un gradient thermique moyen $\nabla T_{vm}$ orienté vers l'équateur, et donc un vent thermique moyen d'ouest (orienté vers l'est). Si on suppose que, en moyenne, les lignes isohypses des surfaces isobares sont purement zonales, alors le vent géostrophique est zonal également. **L'existence d'un vent thermique moyen d'Ouest suggère donc un accroissement des vents d'Ouest dans la troposphère, et un affaiblissement des vents d'Est** dans celle-ci. 

Dans la stratosphère, où le gradient thermique est renversé en été (orienté vers les pôles), les vents d'Ouest typiques des moyennes latitudes s'affaiblissent voire deviennent des vents d'Est (surtout en été lorsque l'insolation aux pôles est maximale); les vents d'Est typiquement trouvés à l'équateur s'intensifient dans la stratosphère.

#### 7.3.3.2 - Décalage des centres dépressionnaires et anticycloniques avec l'altitude

Considérons la situation suivante: un centre dépressionnaire en surface (1000 hPa) est traversé par trois isothermes de la couche comprise entre la surface et l'isobare 700 hPa.

### 7.3.4 - Ecarts entre le vent géostrophique et le vent synoptique réel

La théorie du vent géostrophique développée jusqu'à présent repose sur une hypothèse forte: celle de la stationnarité du champ de vitesse en raison de la relative petitesse, à l'échelle synoptique, du terme d'accélération $\frac{\textrm{d}\vec{u}}{\textrm{d}t}$ dans l'équation du mouvement. Afin d'affiner notre compréhension de la dynamique atmosphérique, nous allons maintenant nous intéresser à des solutions plus réalistes à l'équation du mouvement dans le plan horizontal.

Définissons alors le vecteur égal à la différence entre le vent réel dans l'horizontale $\vec{u}_h$ et le vent géostrophique horizontal $\vec{u}^g_h$ (qui reste pour le moment notre meilleur modèle), et appelons-le par commodité "vent agéostrophique" pour bien signifier qu'il représente tout ce que la géostrophie ne permet pas de rendre compte:

$$
\vec{u}_h^{ag} := \vec{u}_h - \vec{u}_h^g \qquad \textrm{(Vent agéostrophique)}
$$

L'équation du mouvement selon l'horizontale, hors zones de frottement (voir Section 7.3) donne donc:

$$
\frac{\textrm{d}\vec{u}_h}{\textrm{d}t} = - f\vec{k}\times\vec{u}_h - \nabla_p \Phi \Leftrightarrow \frac{\textrm{d}\vec{u}_h}{\textrm{d}t} = - f\vec{k}\times\vec{u}_h^{ag}  - f\vec{k}\times\vec{u}_h^g - \nabla_p \Phi
$$

Or, par définition, le vent géostrophique satisfait la relation

$$
0 = - f\vec{k}\times\vec{u}_h^g - \nabla_p \Phi
$$

ce qui implique donc:

$$
\frac{\textrm{d}\vec{u}_h}{\textrm{d}t} = - f\vec{k}\times\vec{u}_h^{ag} 
$$

En d'autres mots, l'accélération du vent réel (et donc la manière dont il change en direction et/ou en module) est fonction du désaccord entre le vent géostrophique et ce même vent réel. Connaissant la distribution du champ de géopotentiel pour un niveau de pression donné (et donc le champ de vent géostrophique) et le champ de vent réel, nous pouvons donc savoir exactement comment va évoluer ce vent réel. C'est précisément l'idée qu'eut le mathématicien, météorologue et psychologue britannique Lewis Fry Richardson (1881-1953) en 1921-1922, lorsqu'il imagina faire résoudre l'équation ci-dessous dans une sphère géante peuplée par de braves âmes prêtes à calculer. Ce projet ne fut jamais concrétisé, mais préfigura le travail des premiers ordinateurs qui allait arriver trois décennies plus tard.

Notre ambition est maintenant d'apprendre à mieux connaître le vent réel. Nous relaxons l'hypothèse $\frac{\textrm{d}\vec{u}_h}{\textrm{d}t} = 0$ en supposant que la dérivée matérielle du vent réel est, au premier ordre, identique à la dérivée matérielle. Nous posons donc

$$
\frac{\textrm{d}\vec{u}_h}{\textrm{d}t} = \frac{\textrm{d}\vec{u}_h^g}{\textrm{d}t}
$$

On démontre au cours que la composante agéostrophique du vent, $\vec{u}_h^{ag}$ peut se décomposer en deux parties: le vent dit "isallobarique" car évoluant de concert avec les variations temporelles de géopotentiel et dicté par les tendances du champ de pression; et un terme de vent dicté par la répartion spatiale du champ de géopotentiel:

$$
\vec{u}_h^{ag} = \vec{u}_h^{i} + \vec{u}_h^{s}
$$



#### 7.3.4.1 Le vent isallobarique
La première manière d'améliorer notre connaissance du vent réel sur base du champ de géopotentiel est de considérer les fluctuations temporelles de ce dernier: la non-permanence du champ de géopotentiel doit, en toute logique par géostrophie, induire une non-permanence dans le vent réel.

On a démontré au cours la relation
$$
\vec{u}_h^{i} = -\frac{1}{f^2}\nabla_p \frac{\partial{\Phi}}{\partial t} \qquad \textrm{Vent isallobarique / correction isallobarique du vent}
$$

Cette relation exprime la réalité qu'une première correction à apporter au vent géostrophique pour obtenir le vent réel est de prendre en compte la distribution spatiale de la tendance du géopotentiel. 


Examinons maintenant ce concept de vent isallobarique à travers un cas très précis présentant une tendance temporelle marquée: celui de la tempête Lothar qui a touché l'Europe de l'Ouest le 26 décembre 1999.

#### 7.3.4.2 Le vent agéostrophique advectif

La seconde contribution au vent agéostrophique est la correction 

$$
\left\{
\begin{aligned}
u_h^{s} & = & -\frac{1}{f} \left[ u_h^g \left(\frac{\partial v_h^g}{\partial x}\right)_p + v_h^g \left(\frac{\partial v_h^g}{\partial y}\right)_p \right]  \\
v_h^{s} & = & \frac{1}{f} \left[ u_h^g \left(\frac{\partial u_h^g}{\partial x}\right)_p + v_h^g \left(\frac{\partial u_h^g}{\partial y}\right)_p \right]  \\
\end{aligned}
\right.
$$

qui traduit l'influence de termes liés uniquement à la distribution spatiale du vent géostrophique, et donc du géopotentiel.

Effectuons, par commodité, une rotation du système d'axes (originellement orienté est-nord) afin d'être aligné sur les lignes d'égal géopotentiel en ce point, et donc sur le vent géostrophique en ce point. Par convention, on laisse les bas géopotentiels sur notre gauche dans ce nouveau système d'axes: les $x$ positifs pointent donc désormais dans le sens du vent géostrophique, et les $y$ positifs pointent sur sa gauche.

<img style="width:30%" src="./correction_structure/Diapositive1.png">

L'expression ci-dessus se simplifie car $v_h^g$ est par définition nul dans ce nouveau système. On a:
$$
\left\{
\begin{aligned}
u_h^{s} & = & -\frac{1}{f} \left[ u_h^g \left(\frac{\partial v_h^g}{\partial x}\right)_p  \right]  \\
v_h^{s} & = & \frac{1}{f} \left[ u_h^g \left(\frac{\partial u_h^g}{\partial x}\right)_p  \right]  \\
\end{aligned}
\right.
$$


Occupons-nous d'abord de la seconde composante de la correction qui agit donc perpendiculairement/transversalement aux lignes de courant géostrophiques:

$$
v_h^{s}  =  \frac{1}{f} \left[ u_h^g \left(\frac{\partial u_h^g}{\partial x}\right)_p  \right]
$$

En conséquence de cette relation, la correction _transversale_ ($v_h^s$) au vent géostrophique à apporter pour rendre compte de la structure du géopotentiel, est directement proportionnelle au caractère divergent du vent géostrophique $\left(\frac{\partial v_h^g}{\partial x}\right)_p$.

Examinons deux cas:

1. Dans des zones dites "de confluence" (où les isohypses se resserrent), le vent géostrophique croît en intensité avec les $x$ positifs car les lignes isohypses se rapprochent l'une de l'autre. La correction transerversale à apporter au vent géostrophique est donc positive: celui a une trajectoire
2. Dans des zones dites "de diffluence" (où les isohypses s'écartent), le vent géostrophique diminue en intensité avec les $x$ positifs car les lignes isohypses s'éloignent l'une de l'autre. La correction transversale à apporter au vent géostrophique est donc négative.

**Le vent corrigé aura donc une trajectoire moins accentuée que le vent géostrophique autour des zones de bas géopotentiel: il traversera les isohypses en se dirigeant vers les bas géopotentiels dans des zones de confluence, et vers les hauts géopotentiels dans les zones de diffluence. Les trajectoires seront aussi moins accentuées autour des zones de haut géopotentiel, et dans l'hémisphère sud (exercice à démontrer).


<img style="width:50%" src="./correction_structure/Diapositive2.png">




Occupons enfin de la première composante de la correction qui agit donc _longitudinalement_ aux lignes de courant géostrophiques: 
$$
u_h^{s}  =  -\frac{1}{f} \left[ u_h^g \left(\frac{\partial v_h^g}{\partial x}\right)_p  \right]  \\
$$

Dans les zones cycloniques où l'accélération du vent géostrophique est orientée positivement (car dirigée vers le centre du mouvement), la correction longitudinale est négative: le vent corrigé est plus lent que le vent géostrophique.

Dans les zones anticycloniques, la correction longitudinale est positive: le vent corrigé est alors plus rapide que le vent géostrophique.

En résumé, le vent géostrophique tend à sur-estimer les vitesses autour des zones de bas géopotentiel, et à les sous-estimer autour des zones de haut géopotentiel.

## 7.4 - Les mouvements synoptiques verticaux

Notre analyse a jusqu'ici occulté la nature des déplacements synoptiques verticaux. Malgré la faible vitesse verticale des parcelles d'air (de l'ordre de 1-2 cm/s, soit -100 à -200 hPa/jour dans la basse atmosphère), ces mouvements sont pourtant durables et associés à des transformations thermodynamiques souvent décisives pour la prévision météorologique. Il apparaît donc nécessaire de nous intéresser à ces mouvements verticaux à l'échelle synoptique.

### 7.4.1 - L'équation de continuité

Au Chapitre 6, nous avons étudié plusieurs mouvements plans élémentaires. La divergence du champ de vitesse,

$$
\textrm{div } \vec{u} \textrm{ (ou } \nabla \cdot \vec{u} \textrm{)} := \frac{\partial u}{\partial x} + \frac{\partial v}{\partial y} + \frac{\partial w}{\partial z}
$$

quantifie le mouvement de divergence au sein de fluide. On a démontré que, physiquement, la divergence mesure l'accroissement relatif d'une petite surface (en 2D) ou d'un petit volume (en 3D) par unité de temps.

On démontre au cours la relation suivante:

$$
 \frac{1}{\rho} \frac{\textrm{d}\rho}{\textrm{d}t} = - \textrm{div }\vec{u} \ \Leftrightarrow \ \frac{\textrm{d}\rho}{\textrm{d}t} + \rho . \textrm{div }\vec{u}  = 0 \ \Leftrightarrow \   \frac{\partial \rho}{\partial t} + \textrm{div }(\rho \vec{u}) = 0 \qquad \textrm{(Equation de continuité)}
$$

Si nous distinguons maintenant les dimensions horizontales de la dimension verticale,

$$
\textrm{div }\vec{u} = \underbrace{\frac{\partial u}{\partial x} + \frac{\partial v}{\partial y}}_{\textrm{div}_h \vec{u}_h } + \frac{\partial w}{\partial z}
$$

alors l'équation de continuité devient intéressante puisqu'elle permet (indirectement) d'avoir accès à $w$ via l'égalité

$$
\frac{1}{\rho} \frac{\textrm{d}\rho}{\textrm{d}t} + \textrm{div}_h \vec{u}_h +\frac{\partial w}{\partial z} = 0
$$


Cette expression reste néanmoins compliquée, et suppose de pouvoir facilement mesurer les termes dépendant de la densité $\rho$.

On démontre au cours qu'en coordonnées $(t, x, y, p)$ (pression), l'équation de continuité se résume à

$$
\frac{\partial \omega}{\partial p} = - \textrm{div}_h \vec{u}_h
$$

Cette relation traduit le fait qu'**en coordonnées pression**, l'accélération verticale est dictée uniquement par la divergence du champ de vitesse horizontal. Il n'est donc **plus nécessaire de connaître l'évolution du champ de densité** pour connaître l'accélération verticale!

On voit au cours l'inteprétation physique de cette relation fondamentale.

### 7.5.1 - Résumé: les différents types de vents rencontrés
* Vent synoptique réel: vent observé et moyenné sur des échelles synoptiques (~100-1000 km, 1-2 jours)

            * Correction transversale (diffluence/confluence)

```python
# Chargement des données à fréquence 10 minutes dans le dossier data
import os
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
import seaborn as sns
# Définir le chemin du dossier contenant les fichiers CSV
file = 'data/aws_10min.csv'

# Charger le data frame
df = pd.read_csv(file)

# Conversion du time stamp en datetime
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Extraction du vent à 10 m
wind = df[["code", "timestamp", "wind_speed_10m"]]

# Garder uniquement les données validées
import json

def is_valid(qc_string):
    qc = json.loads(qc_string)
    return qc["validated"]["WIND_SPEED_10M"] == True

df["wind_valid"] = df["qc_flags"].apply(is_valid)

wind = df[df["wind_valid"] == True]
wind = wind[["code", "timestamp", "wind_speed_10m"]]

# Affichage de la distribution du vent à 10 m
plt.figure(figsize=(10, 6))
sns.histplot(wind["wind_speed_10m"], bins=30, kde=True)
plt.title("Distribution du vent à 10 m")
plt.xlabel("Vitesse du vent (m/s)")
plt.ylabel("Fréquence")
plt.show()  

# Print des 10 stations contenant le plus de données de vent validées
top_stations = wind["code"].value_counts().head(10)
print(top_stations)
```

```python

# Timeseries des vitesses du vent pour la station la plus représentée et pour l'année 2015.
# Plot avec des points pour ne pas relier des données distantes dans le temps
top_station_code = top_stations.index[0]
top_station_data = wind[(wind["code"] == top_station_code) & (wind["timestamp"].dt.year.isin([2015, 2016]))]
plt.figure(figsize=(9, 3), dpi = 200)
sns.scatterplot(x="timestamp", y="wind_speed_10m", s=2, data=top_station_data)
# Date en format jour-mois-année
plt.gca().xaxis.set_major_formatter(plt.matplotlib.dates.DateFormatter('%d-%b-%Y'))
plt.title(f"Vitesse du vent à 10 m pour la station {top_station_code} en 2015 et 2016")
plt.xlabel("Date")
plt.ylabel("Vitesse du vent (m/s)")
plt.show()  
```

```python

# Analyse en fréquence grâce à une FFT pour la station la plus représentée et pour ces données
from scipy.fft import fft, fftfreq
import numpy as np
# Filtrer les données pour la station la plus représentée

top_station_data = wind[wind["code"] == top_station_code]
# Trier les données par timestamp
top_station_data = top_station_data.sort_values("timestamp")
# Extraire les vitesses du vent et les timestamps
wind_speeds = top_station_data["wind_speed_10m"].values
timestamps = top_station_data["timestamp"].values
# Calculer les intervalles de temps en secondes
time_deltas = (timestamps[1:] - timestamps[:-1]) / pd.Timedelta(seconds=1)
# Vérifier que les intervalles de temps sont constants
if not (time_deltas == time_deltas[0]).all():
    print("Les intervalles de temps ne sont pas constants. La FFT peut ne pas être fiable.")
# Calculer la fréquence d'échantillonnage
sampling_rate = 1 / time_deltas[0]  # en Hz
# Appliquer la FFT
N = len(wind_speeds)
yf = fft(wind_speeds)
xf = fftfreq(N, 1 / sampling_rate)
# Afficher le spectre de fréquences
plt.figure(figsize=(6, 3), dpi = 150)

# On utilise un log plot en x et y pour mieux visualiser les basses fréquences
plt.xscale('log')   
plt.yscale('log')
plt.plot(xf[:N//2], 2.0/N * np.abs(yf[:N//2]))
# On ajoute les ticks correspondant à 10 min, 1 h, 6 h, 12 h, 1 j, 7 j, 30 j
ticks = [1 / 600, 1/ 3600, 1/ 21600, 1/ 43200, 1/ 86400, 1/ 604800, 1/ 2592000]  # en Hz
plt.xticks(ticks, ['10 min', '1 h', '6 h', '12 h', '1 j', '7 j', '30 j'])
plt.title(f"Spectre de fréquences pour la station {top_station_code}")
# Exprimer la fréquence en périodes
plt.xlabel("Période (s)")
plt.ylim(0, 1)

plt.ylabel("Amplitude")

plt.grid()
plt.show()  
```

Le spectre est typique d'un **bruit rouge**, avec une puissance marquée dans les basses fréquences et une lente décroissance vers les fréquences élevées.
